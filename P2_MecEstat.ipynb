```python
def maxwell_distribution(v, m, T):
    """Calcula a distribuição de Maxwell-Boltzmann f(v) com valores reais."""
    
    #Boltzmann constant (J/K)
    k_B = 1.380649e-23  
    
    #Normalization
    constante = 4.0 * np.pi * (m / (2.0 * np.pi * k_B * T))**1.5
    
    #Distribution
    f_v = constante * (v**2) * np.exp(-m * v**2 / (2.0 * k_B * T))
    return f_v
```

```python


m = 4.65e-26 #Mass of N_2 molecule in kg
v = np.linspace(0, 2000, 1000) 

#Temperatures
T1 = 273.15  # 0 °C
T2 = 373.15  # 100 °C
T3 = 773.15  # 500 °C
# Distributions
f_T1 = maxwell_distribution(v, m, T1)
f_T2 = maxwell_distribution(v, m, T2)
f_T3 = maxwell_distribution(v, m, T3)


plt.figure(figsize=(10, 6)) 


plt.plot(v, f_T1, color = sns.color_palette("Paired")[1],  label=f"T = {T1} (K)")
plt.plot(v, f_T2, color = sns.color_palette("Paired")[3], label=f"T = {T2} (K)")
plt.plot(v, f_T3, color = sns.color_palette("Paired")[5], label=f"T = {T3} (K)")


plt.title("Maxwell-Boltzmann distribution $f(v)$", fontsize=16)
plt.xlabel("$ v$ [m/s]", fontsize=12)
plt.ylabel("$f(v)$", fontsize=12)
plt.legend(fontsize=12) 
#plt.grid(True) 


plt.xlim(left=0) 
plt.ylim(bottom=0) 


plt.savefig('V_dist.png')
plt.show()
```
